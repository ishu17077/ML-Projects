```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Lasso
from sklearn.linear_model import LinearRegression
from sklearn import metrics
```

```python
car_dataset = pd.read_csv("./car data.csv")
```

## Encoding the categorical data

```python
# encoding Fuel_Type Column
car_dataset.replace({'Fuel_Type': {'Petrol': 0, 'Diesel': 1, 'CNG': 2}}, inplace=True)
car_dataset.replace({'Seller_Type':{'Dealer': 0, 'Individual': 1}}, inplace=True)
car_dataset.replace({'Transmission': {'Manual': 0, 'Automatic': 1}}, inplace=True)
```

## Splitting the data into training and split data

```python
X = car_dataset.drop(['Car_Name', 'Selling_Price'], axis=1)
Y = car_dataset[['Selling_Price']]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.1, random_state=2)
```

## Model Training

### 1. Linear Regression

```python
lin_reg_model = LinearRegression()
```

```python
lin_reg_model.fit(X_train, Y_train)
```

```python
# predict test data
test_data_pred = lin_reg_model.predict(X_test)
```

```python
# R square error
error_score = metrics.r2_score(Y_test, test_data_pred)
print("R squared Error: ", error_score)
```

## Visualize Predicted prices vs Actual

```python
plt.style.use('dark_background')
plt.scatter(Y_test, test_data_pred, cmap='viridis', alpha=0.7)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual Price vs Predicted")
plt.show()
```


### 2. Lasso Regression (For data that are not directly correlated, correlated data all -> Linear else this)

```python
lasso_reg_model = Lasso()
```

```python
lasso_reg_model.fit(X_train, Y_train)
```

```python
test_data_pred = lasso_reg_model.predict(X_test)
```

```python
error_score = metrics.r2_score(Y_test, test_data_pred)
print("R squared error: ", error_score)
```

```python
plt.style.use('dark_background')
plt.scatter(Y_test, test_data_pred)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual Price vs Predicted Price")
plt.show()
```
